# Shared Variables and Independent Cells

Use the Shared / Independent menu in the Workspace header. Changing mode asks to restart Python; code and outputs stay in place, and live variables are cleared. Rerun the code cells after changing mode.

**Shared Variables:** cells and console in this project use the same variable names. After Alpha runs, Beta can see `alpha_only`; assigning `x` in Beta replaces the project's `x`.

**Independent Cells:** each cell retains its own names. Alpha's `x` and Beta's `x` appear separately in Workspace. Beta cannot see `alpha_only`. Imports and functions belong to their cell too. The console uses the selected cell.

Run Alpha and Beta, then select each cell and type `print(x)` in the console. Independent mode prints 10 for Alpha and 20 for Beta; shared mode prints the latest assigned value.
