# Live plots
Run the next cell to watch one figure update in place for five seconds. Open Figures or the inline Open Figure control to view the live figure at a larger size. The code editor keeps its space. Stop interrupts the loop, leaving the last frame. Save/reopen retains the last frame; run again to start a new live display.

```python
import time
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import display

x = np.linspace(0, 4 * np.pi, 500)
fig, ax = plt.subplots(figsize=(9, 4), layout="constrained")
line, = ax.plot(x, np.sin(x), color="#007acc", linewidth=2)
ax.set(xlabel="Position", ylabel="Amplitude", ylim=(-1.2, 1.2), title="Live wave · frame 0 / 50")
ax.grid(alpha=0.2)
try:
    view = display(fig, display_id=True)
    for frame in range(1, 51):
        line.set_ydata(np.sin(x - frame * 0.15))
        ax.set_title(f"Live wave · frame {frame} / 50")
        view.update(fig)
        time.sleep(0.1)
finally:
    plt.close(fig)
```
