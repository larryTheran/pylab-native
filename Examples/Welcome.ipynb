```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
plt.style.use('dark_background')
plt.rcParams.update({'figure.facecolor': '#1e1e1e', 'axes.facecolor': '#1e1e1e'})

x = np.linspace(0, 2 * np.pi, 120)
signal = np.column_stack([x, np.sin(x)])
measurements = pd.DataFrame({"time": x[:10], "amplitude": np.sin(x[:10])})
signal[:5]
```

```python
plt.figure(figsize=(7, 3))
plt.plot(x, np.sin(x), color="#569cd6", linewidth=2)
plt.title("A shared scientific workspace")
plt.xlabel("Time")
plt.ylabel("Amplitude")
plt.grid(alpha=0.2)
plt.tight_layout()
plt.show()
```
