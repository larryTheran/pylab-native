```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)
time = np.linspace(0, 2, 400)
signal = np.sin(2 * np.pi * 3 * time) + rng.normal(0, 0.12, time.size)
measurements = pd.DataFrame({"time_s": time, "voltage_V": signal})
print(f"Loaded {len(measurements)} measurements into the shared project workspace.")
```

```python
def moving_average(values, width=11):
    weights = np.ones(width) / width
    return np.convolve(values, weights, mode="same")

smoothed = moving_average(signal)
summary = measurements.describe()
print(summary)
# Right-click signal or measurements in Workspace > Inspect Statistics.
# Double-click measurements to open the table editor.
```

```python
plt.figure(figsize=(8, 4))
plt.plot(time, signal, alpha=0.35, label="Measured")
plt.plot(time, smoothed, linewidth=2, label="Smoothed")
plt.xlabel("Time (s)")
plt.ylabel("Voltage (V)")
plt.title("Shared data across named notebook cells")
plt.grid(alpha=0.25)
plt.legend()
plt.tight_layout()
plt.show()
# This plot also appears under Figures, including plots run in the console.
```
